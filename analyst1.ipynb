```python
import pandas as pd

df = pd.read_csv("customer_shopping_behavior.csv")

df.head() #first 5 rows of the dataset
```

```python
df.describe(include='all')
```

```python
df.isnull().sum() #isnull() returns true if null, then .sum() add that, so we count the number of null values.
```

```python
df['Review Rating']= df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))
#find the missing values and then fill it with median values
```

```python
df.isnull().sum() # first find the null values, and then sum of that values
```

```python
original_columns=df.columns.copy()
original_columns
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df.columns
```

```python
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
df.columns
```

```python
#create a new column age_group
labels=['young adult','adult','middle-aged','senior']
df['age_group']= pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
#create column purchase frequency days
frequency_mapping={
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90
}
df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
df[['discount_applied','promo_code_used']].head(10)
```

```python
df['promo_code_used']=df['promo_code_used'].astype(str)# .astype convert into different data types
```

```python
(df['discount_applied'] == df['promo_code_used']).all() # .all returns true if every values matches
```

```python
#dropping promo_code_used as it is redundant
df=df.drop('promo_code_used',axis=1)
```
